# How is programming used in physics?



For example, a classical problem in PHYS 206 Mechanics is the free fall of a ball under gravity. Programming can be used to simulate the motion, calculate the trajectory, and visualize the results, at different initial conditions. Once the code is written correctly, there is no need to manually redo the calculations for each new scenario; the program can simply be run with the updated parameters without any human errors. 


The code for simulating free fall might looks something like this:

```python
import matplotlib.pyplot as plt
x = 10 #
v = 0 # initial velocity
g = 9.8 # acceleration due to gravity
dt = 0.01 # time step
t = 0 # initial time
positions = [] # to store positions over time
velocities = [] # to store velocities over time
time = [] # to store time over time
while x > 0:
    v = v - g * dt
    x = x + v * dt
    t = t + dt
    positions.append(x)
    velocities.append(v)
    time.append(t)
fig, ax = plt.subplots(figsize=(3,3))
plt.plot(time, positions)
plt.xlabel('Time (s)')
plt.ylabel('Position (m)')
```

## Conway's Game of Life

A baby universe of cells evolving according to simple principles.

- Any live cell with fewer than two live neighbors dies, as if by underpopulation.
- Any live cell with two or three live neighbors lives on to the next generation.
- Any live cell with more than three live neighbors dies, as if by overpopulation.
- Any dead cell with exactly three live neighbors becomes a live cell, as if by reproduction.

### Schrodinger's Cat

For the simplest system, $|\psi\rangle$ is a 2 dimensional vector (2 numbers) that represents two states (live and dead). However $N$ such systems are described by $2^N$ numbers! For $N=100$, it is about $1.2\times 10^{30}$.
