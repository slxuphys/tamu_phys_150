## Jupyter Notebook

In this course, we will use Jupyter Notebook as our primary programming environment. Jupyter Notebook is an open-source web application that allows you to create and share documents that contain live code, equations, visualizations, and narrative text. It supports many programming languages, including Python, which we will use in this course.

We will Jupyter Notebook for lecture notes, assignments, group projects, and finals. 

There are tree ways to open and edit a Jupyter Notebook:

1. Use [Google Colab](https://colab.research.google.com/) to open the notebook directly in your browser. This is the easiest way to get started, as it does not require any installation. When you create a notebook in Google Colab, it will automatically save your work to your Google Drive under `My Drive/Colab Notebooks`. You can also download the notebook as a `.ipynb` file.
2. Use [Anaconda](https://www.anaconda.com/download/success) to install Python and Jupyter Notebook on your local machine. Then you can open the notebook in your browser by running the app `jupyter notebook` or open anaconda navigator and launch the notebook from there. 
3. After installing Anaconda, you can also download and use [VS Code](https://code.visualstudio.com/) to open and edit the notebook. This is more recommended. When you install VS Code and open a jupyter notebook for the first time, it will ask you to install several extensions. Please install all of them. You can also install the extensions manually by searching for "Python" and "Jupyter" in the Extensions tab.

When you see a colab badge  [![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/slxuphys/tamu_phys_150/blob/master/2026/lecture%2001%20introduction.ipynb) in the lecture notes, you can click on it to open the notebook in Google Colab. You can also open the notebook in VS Code or Anaconda by downloading the `.ipynb` file from GitHub.
