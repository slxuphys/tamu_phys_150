# Statements

- Python interprets and executes statements line by line.
- A statement is a complete instruction that the Python interpreter can execute.
- Examples of statements include assignment statements, print statements, and control flow statements like if and for.


````python
print("Hello, World!")
print("Second statement")
a = 5 + 2
print(a)
````

- Python is case sensitive. (print and Print mean different things)
- Python uses the `#` symbol for comments. Comments are ignored by the interpreter and are used to explain the code.
- One statement per line is a common practice in Python to ensure readability and maintainability of the code.
    - Python supports multi-statement lines using the semicolon `;` to separate multiple statements on a single line.
    - Python allows line continuation using the backslash `\` character to split a long statement across multiple lines.

````python
Print("Hello, World!") # This will cause an error because Python is case sensitive and the correct function is `print`
````

````python
a=5 b=2 # This will cause an error because Python expects one statement per line or a semicolon to separate multiple statements
````

````python
a=5;b=2
print(a, b)
````

````python
# Using line continuation with the backslash character to split a long statement across multiple lines
a = 1*2*3\
*4*5*6
print(a)
````

- Indentation is used to define the scope of loops, functions, and conditionals in Python. It is important to use consistent indentation (four spaces) throughout your code.
- Do not indent unless you are inside a loop, function, or conditional block.

````python
print("Hello, World!") 
  print("This line is indented.")
````
