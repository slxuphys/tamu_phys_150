# Why Do We Need Variables?

In principle, we could write out every single value explicitly in our code, but this would be highly inefficient and error-prone. Variables allow us to store and manipulate data dynamically, making our programs more flexible and easier to maintain.

Example:

```python
# Without variables
print(5 + 3)
print(5 * 3)

# With variables
a = 5
b = 3
print(a + b)
print(a * b)
```



### A Physics Example

Suppose a car starts with an initial velocity of $v_0=0\,\mathrm{m/s}$ and accelerates at $a=2\,\mathrm{m/s^2}$ for $t=5\,\mathrm{s}$. Its displacement is

$$x=v_0t+\frac{1}{2}at^2$$



Without variables, we can calculate the answer for this one case:

```python
print(0 * 5 + 0.5 * 2 * 5**2)
```

This works, but it is difficult to read and reuse. If any value changes, we must find and replace the numbers in the formula.



With variables, the code matches the physics equation:

```python
v_0 = 0       # m/s
a = 2         # m/s^2
t = 5          # s

x = v_0 * t + 0.5 * a * t**2
print(x)      # m
```

The variables $v_0$, $a$, $t$, and $x$ represent the physical quantities in the equation. We can now change the initial velocity, acceleration, or time without rewriting the formula.

### String Indexing and Slicing

How can you extract the year, month, and day from the user's input, `mm/dd/yyyy`?

- Single character:
  ```python
  s[index]
  ```

  A negative index means `length + negative index`.

- Substring:
  ```python
  s[start:end:step]
  ```

  `start`: the starting index for the substring.

  `end`: the ending index for the substring __+1__.

  `step`: can be positive or negative.

- This can be confusing at first, so use trial and error.

````python
s='ABCDEF' #the length of the string is 10
print(s[1]) # single number to access a single character;
#the index starts at 0
````

````python
#negative index means length + negative index
print(s[-1]) #last character
#same as
print(s[-1+len(s)])
````

````python
# access a substring
# two colons;three numbers
s='0123456789'
print(s[1:5:1]) #index from 1 to 4 (not 5) with step 1
print(s[1:5:2]) #every other number from index 1 to 4
````

#### Simplified Notation
- You can omit an index (the start or end) if it is at the beginning or end of the string.
- You can omit the step if it is 1.

````python
s='ABCDEFGHIJK'
print(s[1:5]) #if the step is one, the last column and number can be omitted
print(s[:5]) #if starting from the beginning, the first number can be omitted
print(s[1:]) #if going all the way to end, the second number can be omitted


# step can be -negative
print(s[5:0:-1]) #index frm 5 to 1(!), not 0
print(s[5::-1]) #index from 5 to 0; the second number is omitted
print(s[::-1]) #the whole string to reverse order

# negative index should be interpreted as length + negative index
# useful if we do not know the length
print(s[len(s)-5:len(s):1]) #last 5 characters; most formal method
print(s[-5:]) #the length '10' is omitted because -5 is a negative index
# the end index is omitted because it goes only the way to the end
# the step is omitted because the step is 1.
````
