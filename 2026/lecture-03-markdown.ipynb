# Jupyter Notebook
From now on we will use jupyter notebook to write our notes, codes, homework and exams.

It allows you to create and share documents that contain live code, equations, visualizations, and narrative text. It is widely used in data science, scientific computing, and machine learning for its interactive and user-friendly interface.

Any file with the extension `.ipynb` is a Jupyter Notebook file, such as this one.

## Cell Types
There are two main types of cells in a Jupyter Notebook: code cells and markdown cells.
- Code Cells (for coding): Code cells are indicated by a `In [ ]:` prompt or a running button on the left side. 
- Markdown Cells (for documentation):  no `In [ ]:` prompt nor a running button.

- running a cell: press `Shift + Enter` to run the current cell and move to the next one.

This cell is a markdown cell. The cell below is a code cell. 

```python
## this is a code cell. Running code successfully should print "hello world!" below, indicating that python is working.
print('hello world!')
```

## Basic Control

- editing mode: click a code cell or double click a markdown cell to edit it.
- command mode: press `Esc` to enter command mode, where you can perform notebook-level actions such as adding or deleting cells.



In command mode, you can use keyboard shortcuts to perform various actions on the notebook cells.

- adding a new cell: press `A` to add a cell above the current cell or `B` to add a cell below the current cell.
- deleting a cell: press `D` twice (`DD`) to delete the current cell. (ctrl+M D on google colab)
- converting a cell: press `Y` (ctrl+M Y on colab) to convert the current cell to a code cell or `M` (ctrl+M Y on colab) to convert it to a markdown cell.

### Display equations

### Display equation with more lines
Some equations are too long to fit in one line. You can use the `aligned` environment to align multiple lines of equations.

\begin{aligned}
